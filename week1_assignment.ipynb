```python
#importing important libraries
import numpy as np
import pandas as pd
pd.set_option('display.width', 1000)
pd.set_option('display.max_columns', None)
```

```python
#loading dataset
df = pd.read_csv('/Users/HP/Downloads/student_marks_dataset/StudentsPerformance.csv')
print(df.head())
```

```python
#checking for missing values
print('missing values in each columns')
df.isnull().sum()
```

```python
#checking duplicates
print('Duplicates columns')
df.duplicated()
```

```python
#cleaning data by filling null values
df['race/ethnicity'].fillna('Unknown')
```

```python
#again checking for null values
df.isnull().sum()
```

```python
#calculating total sore and average score
df['total score'] = df['math score']+ df['reading score']+ df['writing score']
df['average score'] = df['total score']/3
df.head()
```

```python
#calculating minimum and maximum score
subjects = ['math score', 'reading score', 'writing score']
df['max score'] = df[subjects].max(axis=1)
df['min score'] = df[subjects].min(axis=1)
df.head()
```
